## 3. Resumo teórico
A simulação ajuda a visualizar o comportamento das grandezas físicas e facilita interpretação de questões.

## 4. Fórmulas principais
- **Q = m·c·ΔT**
- Use unidades: cal ou J.

## 5. Exemplo resolvido
Substitua valores simples na fórmula e confirme o resultado no gráfico.


```python
import numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets

def simular(m,c,deltaT):
    Q=m*c*deltaT
    temps=np.linspace(0,deltaT,100)
    q=m*c*temps
    plt.figure(figsize=(8,4)); plt.plot(temps,q,lw=2)
    plt.xlabel('ΔT (°C)'); plt.ylabel('Q (cal)'); plt.grid(); plt.show()
    print(f'Calor trocado: {Q:.1f} cal')
widgets.interact(simular,m=(100.0,1000.0,50.0),c=(0.1,1.0,0.1),deltaT=(-50.0,80.0,5.0));
```
