## 4. Fórmulas principais
- Fórmula 1 (modelo geral): $G = G_0 + k\cdot t$
- Fórmula 2 (modelo de comparação): $\Delta G = G_f - G_i$

**Grandezas:**
- $G$: grandeza física analisada (unidade depende do tema).
- $G_0$: valor inicial.
- $k$: taxa de variação.
- $t$: tempo em segundos (s).

## 5. Exemplo resolvido
Suponha $G_0=10$ e $k=3$ durante $t=4\,s$.
1. Substitua na expressão: $G = 10 + 3\cdot4$.
2. Calcule: $G = 22$.
3. Interpretação: em 4 s, a grandeza aumentou 12 unidades.

## 6. Simulador interativo em Python


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

# Simulador genérico didático para o tema do notebook.
def simulador(valor_inicial=0.0, taxa=2.0, tempo_total=10.0, pontos=200):
    t = np.linspace(0, tempo_total, int(pontos))
    y = valor_inicial + taxa * t

    plt.figure(figsize=(8,4.5))
    plt.plot(t, y, linewidth=2, label='Grandeza simulada')
    plt.title('Simulação Interativa')
    plt.xlabel('Tempo (s)')
    plt.ylabel('Valor da grandeza (u.a.)')
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.show()

    print(f'Valor inicial = {valor_inicial:.2f}')
    print(f'Taxa de variação = {taxa:.2f} por segundo')
    print(f'Valor final aproximado = {y[-1]:.2f}')

interact(
    simulador,
    valor_inicial=widgets.FloatSlider(value=0, min=-20, max=20, step=0.5, description='Inicial'),
    taxa=widgets.FloatSlider(value=2, min=-10, max=10, step=0.5, description='Taxa'),
    tempo_total=widgets.FloatSlider(value=10, min=1, max=30, step=1, description='Tempo'),
    pontos=widgets.IntSlider(value=200, min=50, max=500, step=50, description='Pontos')
);
```
