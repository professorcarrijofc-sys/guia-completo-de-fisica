## 4. Fórmulas principais
- **Q = m·c·ΔT**
- Grandezas: Q (J), m (kg), c (J/kg°C), ΔT (°C).
- Sempre confira unidades antes de calcular.

## 5. Exemplo resolvido
Água: m=0,5 kg, c=4200, ΔT=10°C.
Q=0,5·4200·10=21000 J.

## 6. Simulador interativo em Python
O código abaixo usa **numpy**, **matplotlib** e **ipywidgets**, compatíveis com Google Colab.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
plt.style.use('seaborn-v0_8-whitegrid')

def sim(m,c,dT):
    q=m*c*dT
    temps=np.linspace(0,dT,120); Q=m*c*temps
    plt.figure(figsize=(8,4)); plt.plot(temps,Q,label='Q = m·c·ΔT')
    plt.title('Calorimetria'); plt.xlabel('Variação de temperatura (°C)'); plt.ylabel('Calor (J)'); plt.legend(); plt.grid(); plt.show()
    print(f'Calor trocado: {q:.2f} J')
widgets.interact(sim,m=(0.1,5.0,0.1),c=(100.0,5000.0,100.0),dT=(-50.0,100.0,5.0));
```
