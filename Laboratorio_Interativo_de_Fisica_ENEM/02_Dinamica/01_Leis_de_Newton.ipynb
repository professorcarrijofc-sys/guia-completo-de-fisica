## 4. Fórmulas principais
- **a = F/m**
- Use unidades: N e kg.

## 5. Exemplo resolvido
Substitua valores simples na fórmula e confirme o resultado no gráfico.


```python
import numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets

def simular(F,m,t_max):
    a=F/m
    t=np.linspace(0,t_max,200)
    v=a*t
    plt.figure(figsize=(8,4)); plt.plot(t,v,'r',lw=2)
    plt.title('Dinâmica: velocidade x tempo (força constante)'); plt.xlabel('Tempo (s)'); plt.ylabel('Velocidade (m/s)'); plt.grid(); plt.show()
    print(f'Aceleração: {a:.2f} m/s²')
widgets.interact(simular,F=(0.0,100.0,1.0),m=(1.0,20.0,0.5),t_max=(1,20,1));
```

## 7. Interpretação física dos resultados
Observe como os gráficos mudam ao variar os controles e relacione com a fórmula.
