## 4. Fórmulas principais
- **x = v0·t e y = (g·t²)/2**
- Grandezas: x (m), y (m), v0 (m/s), t (s), g (m/s²).
- Sempre confira unidades antes de calcular.

## 5. Exemplo resolvido
Uma bola sai de uma mesa com v0=8 m/s e cai por 1,5 s.
Alcance: x=8·1,5=12 m.
Queda: y=(10·2,25)/2=11,25 m.

## 6. Simulador interativo em Python
O código abaixo usa **numpy**, **matplotlib** e **ipywidgets**, compatíveis com Google Colab.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
plt.style.use('seaborn-v0_8-whitegrid')

def sim(v0,h0,g):
    t_queda=np.sqrt(2*h0/g); t=np.linspace(0,t_queda,250); x=v0*t; y=h0-0.5*g*t**2
    plt.figure(figsize=(8,4)); plt.plot(x,y,label='Trajetória')
    plt.title('Lançamento Horizontal'); plt.xlabel('Alcance x (m)'); plt.ylabel('Altura y (m)'); plt.legend(); plt.grid(); plt.show()
    print(f'Alcance: {x[-1]:.2f} m | Tempo de queda: {t_queda:.2f} s')
widgets.interact(sim,v0=(1.0,30.0,1.0),h0=(1.0,50.0,1.0),g=(9.8,10.0,0.2));
```
