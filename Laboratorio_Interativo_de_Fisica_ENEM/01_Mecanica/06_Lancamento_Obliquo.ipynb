## 4. Fórmulas principais
- **vx = v0·cos(θ), vy = v0·sen(θ), alcance ≈ v0²·sen(2θ)/g**
- Grandezas: v0 (m/s), θ (graus), g (m/s²), alcance (m).
- Sempre confira unidades antes de calcular.

## 5. Exemplo resolvido
Para v0=20 m/s, θ=45°, g=10: alcance≈(400·1)/10=40 m.

## 6. Simulador interativo em Python
O código abaixo usa **numpy**, **matplotlib** e **ipywidgets**, compatíveis com Google Colab.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
plt.style.use('seaborn-v0_8-whitegrid')

def sim(v0,angulo,g):
    th=np.radians(angulo); t_voo=2*v0*np.sin(th)/g; t=np.linspace(0,t_voo,250)
    x=v0*np.cos(th)*t; y=v0*np.sin(th)*t-0.5*g*t**2
    plt.figure(figsize=(8,4)); plt.plot(x,y,label='Trajetória')
    plt.title('Lançamento Oblíquo'); plt.xlabel('x (m)'); plt.ylabel('y (m)'); plt.legend(); plt.grid(); plt.show()
    print(f'Alcance: {x[-1]:.2f} m')
widgets.interact(sim,v0=(5.0,50.0,1.0),angulo=(10.0,80.0,1.0),g=(9.8,10.0,0.2));
```
