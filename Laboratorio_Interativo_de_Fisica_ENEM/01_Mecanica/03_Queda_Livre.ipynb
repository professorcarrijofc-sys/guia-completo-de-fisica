## 4. Fórmulas principais
- **v = g·t e h = (g·t²)/2 (saindo do repouso)**
- Grandezas: g (m/s²), t (s), v (m/s), h (m).
- Sempre confira unidades antes de calcular.

## 5. Exemplo resolvido
Uma bola cai por 3 s com g=10 m/s².
v = 10·3 = 30 m/s.
h = (10·9)/2 = 45 m.

## 6. Simulador interativo em Python
O código abaixo usa **numpy**, **matplotlib** e **ipywidgets**, compatíveis com Google Colab.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
plt.style.use('seaborn-v0_8-whitegrid')

def sim(g,tmax):
    t=np.linspace(0,tmax,200); h=0.5*g*t**2; v=g*t
    plt.figure(figsize=(8,4)); plt.plot(t,h,label='h(t)'); plt.plot(t,v,label='v(t)')
    plt.title('Queda Livre'); plt.xlabel('Tempo (s)'); plt.ylabel('Valor numérico'); plt.legend(); plt.grid(); plt.show()
widgets.interact(sim,g=(9.8,10.0,0.2),tmax=(1,10,1));
```
