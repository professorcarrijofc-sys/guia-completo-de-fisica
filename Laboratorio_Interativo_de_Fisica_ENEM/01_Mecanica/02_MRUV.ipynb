## 4. Fórmulas principais
- **v = v0 + a·t e s = s0 + v0·t + (a·t²)/2**
- Grandezas: v (m/s), v0 (m/s), a (m/s²), t (s), s (m).
- Sempre confira unidades antes de calcular.

## 5. Exemplo resolvido
Um carro parte com v0=2 m/s e a=3 m/s² por 4 s.
v = 2 + 3·4 = 14 m/s.
s = 0 + 2·4 + (3·16)/2 = 8 + 24 = 32 m.

## 6. Simulador interativo em Python
O código abaixo usa **numpy**, **matplotlib** e **ipywidgets**, compatíveis com Google Colab.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
plt.style.use('seaborn-v0_8-whitegrid')

def sim(s0,v0,a,tmax):
    t=np.linspace(0,tmax,200); s=s0+v0*t+0.5*a*t**2; v=v0+a*t
    fig,ax=plt.subplots(1,2,figsize=(12,4))
    ax[0].plot(t,s,label='s(t)'); ax[0].set_title('Posição x Tempo'); ax[0].set_xlabel('t (s)'); ax[0].set_ylabel('s (m)'); ax[0].legend(); ax[0].grid()
    ax[1].plot(t,v,color='orange',label='v(t)'); ax[1].set_title('Velocidade x Tempo'); ax[1].set_xlabel('t (s)'); ax[1].set_ylabel('v (m/s)'); ax[1].legend(); ax[1].grid()
    plt.show()
widgets.interact(sim,s0=(-20.0,20.0,1.0),v0=(-20.0,20.0,1.0),a=(-10.0,10.0,0.5),tmax=(1,20,1));
```
