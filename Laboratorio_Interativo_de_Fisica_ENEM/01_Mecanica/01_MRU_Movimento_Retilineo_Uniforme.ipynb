## 2. Objetivo da atividade
Entender como posição, velocidade e tempo se relacionam no MRU.

## 3. Resumo teórico
No MRU, a velocidade é constante. Isso significa que o móvel percorre distâncias iguais em tempos iguais.

## 4. Fórmulas principais
- **s = s0 + v·t**
- s: posição (m)
- s0: posição inicial (m)
- v: velocidade (m/s)
- t: tempo (s)

## 5. Exemplo resolvido
Um carro parte de s0=10 m com v=5 m/s por 8 s.

s = 10 + 5·8 = 50 m.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def simular_mru(s0, v, t_max):
    t = np.linspace(0, t_max, 200)
    s = s0 + v*t
    plt.figure(figsize=(8,4))
    plt.plot(t, s, color='royalblue', lw=2)
    plt.title('MRU: posição x tempo')
    plt.xlabel('Tempo (s)')
    plt.ylabel('Posição (m)')
    plt.grid(True)
    plt.show()
    print(f'Posição final: {s[-1]:.2f} m')

widgets.interact(simular_mru,
                 s0=widgets.FloatSlider(value=0,min=-50,max=50,step=1,description='s0 (m)'),
                 v=widgets.FloatSlider(value=10,min=-30,max=30,step=1,description='v (m/s)'),
                 t_max=widgets.IntSlider(value=10,min=1,max=30,step=1,description='t máx (s)'));
```

## 7. Interpretação física dos resultados
A inclinação da reta representa a velocidade. Reta crescente: velocidade positiva; decrescente: negativa.

## 8. Erros comuns em provas
- Esquecer a posição inicial.
- Trocar sinal da velocidade.
