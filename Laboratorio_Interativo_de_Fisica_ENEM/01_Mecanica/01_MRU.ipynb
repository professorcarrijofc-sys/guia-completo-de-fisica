## 4. Fórmulas principais
- **s = s0 + v·t**
- Grandezas: s (m), s0 (m), v (m/s), t (s).
- Sempre confira unidades antes de calcular.

## 5. Exemplo resolvido
Um ciclista parte de s0=20 m e mantém v=4 m/s por 15 s.
Passo 1: use s = s0 + v·t.
Passo 2: s = 20 + 4·15 = 80 m.
Resposta: posição final de 80 m.

## 6. Simulador interativo em Python
O código abaixo usa **numpy**, **matplotlib** e **ipywidgets**, compatíveis com Google Colab.


```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
plt.style.use('seaborn-v0_8-whitegrid')

def sim(s0,v,tmax):
    t=np.linspace(0,tmax,200); s=s0+v*t
    plt.figure(figsize=(8,4)); plt.plot(t,s,label='s(t)')
    plt.title('MRU: Posição x Tempo'); plt.xlabel('Tempo (s)'); plt.ylabel('Posição (m)'); plt.legend(); plt.grid(); plt.show()
    print(f'Posição final: {s[-1]:.2f} m')
widgets.interact(sim,s0=(-50.0,50.0,1.0),v=(-30.0,30.0,1.0),tmax=(1,30,1));
```
