## 4. Fórmulas principais
- **U = R·i**
- Use unidades: V, Ω e A.

## 5. Exemplo resolvido
Substitua valores simples na fórmula e confirme o resultado no gráfico.


```python
import numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets

def simular(R,i):
    U=R*i
    i_vals=np.linspace(0,i,100)
    U_vals=R*i_vals
    plt.figure(figsize=(8,4)); plt.plot(i_vals,U_vals,lw=2)
    plt.xlabel('Corrente (A)'); plt.ylabel('Tensão (V)'); plt.grid(); plt.show()
    print(f'Tensão aplicada: {U:.2f} V')
widgets.interact(simular,R=(1.0,100.0,1.0),i=(0.1,10.0,0.1));
```

## 7. Interpretação física dos resultados
