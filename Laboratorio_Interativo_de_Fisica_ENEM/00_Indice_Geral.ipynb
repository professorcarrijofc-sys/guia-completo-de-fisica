# 00_Indice_Geral — Laboratório Interativo de Física ENEM

Bem-vindo(a)! Este índice organiza todos os laboratórios em ordem sugerida para estudo.

## Apresentação
Este material foi criado para alunos do Ensino Médio com foco em ENEM e vestibulares. Cada notebook funciona de forma independente no Google Colab.

## Lista de experimentos
- 01_Mecanica: MRU, MRUV, Queda Livre, Lançamentos.
- 02_Dinamica: 2ª Lei de Newton, Plano Inclinado, Lei de Hooke.
- 03_Energia: Conservação de Energia, Colisões.
- 04_Termologia: Calorimetria, Dilatação, Gases.
- 05_Ondulatoria: Ondas em Cordas, Efeito Doppler.
- 06_Optica: Espelhos, Lentes.
- 07_Eletricidade: Lei de Ohm, Resistores, Circuitos Simples.

## Sugestão de ordem de estudo
1. Mecânica
2. Dinâmica
3. Energia
4. Termologia
5. Ondulatória
6. Óptica
7. Eletricidade

## Instruções gerais de uso
1. Abra um notebook por vez no Colab.
2. Vá em **Ambiente de execução > Executar tudo**.
3. Interaja com sliders e analise gráficos.
4. Faça os exercícios e confira o gabarito comentado.
