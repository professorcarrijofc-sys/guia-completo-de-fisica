# 00_Indice_Geral - Laboratório Interativo de Física ENEM

Bem-vindo(a)! Este laboratório reúne **20 experimentos interativos** para Ensino Médio, ENEM e vestibulares.

## Como usar no Google Colab
1. Abra o notebook desejado.
2. Clique em **Ambiente de execução > Executar tudo**.
3. Use os controles interativos e observe os gráficos.
4. Leia a interpretação e faça os exercícios.
