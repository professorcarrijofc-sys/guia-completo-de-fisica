## 3. Resumo teórico
A simulação ajuda a visualizar o comportamento das grandezas físicas e facilita interpretação de questões.

## 4. Fórmulas principais
- **v = λ·f**
- Use unidades: m e Hz.

## 5. Exemplo resolvido
Substitua valores simples na fórmula e confirme o resultado no gráfico.


```python
import numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets

def simular(lmbd,f):
    v=lmbd*f
    x=np.linspace(0,4*lmbd,400)
    y=np.sin(2*np.pi*x/lmbd)
    plt.figure(figsize=(8,4)); plt.plot(x,y); plt.axhline(0,color='k',lw=1)
    plt.title('Onda senoidal'); plt.xlabel('Posição (m)'); plt.ylabel('Amplitude'); plt.grid(); plt.show()
    print(f'Velocidade da onda: {v:.2f} m/s')
widgets.interact(simular,lmbd=(0.5,10.0,0.5),f=(0.5,20.0,0.5));
```
