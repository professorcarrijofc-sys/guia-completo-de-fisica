## 4. Fórmulas principais
- **Em = Ec + Ep**
- Use unidades: J.

## 5. Exemplo resolvido
Substitua valores simples na fórmula e confirme o resultado no gráfico.


```python
import numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets

def simular(m,h_max):
    g=10
    h=np.linspace(0,h_max,200)
    Ep=m*g*h
    Em=np.full_like(h,m*g*h_max)
    Ec=Em-Ep
    plt.figure(figsize=(8,4)); plt.plot(h,Ep,label='Ep'); plt.plot(h,Ec,label='Ec'); plt.plot(h,Em,'k--',label='Em')
    plt.xlabel('Altura (m)'); plt.ylabel('Energia (J)'); plt.legend(); plt.grid(); plt.show()
widgets.interact(simular,m=(1.0,10.0,0.5),h_max=(1.0,30.0,1.0));
```

## 7. Interpretação física dos resultados
Observe como os gráficos mudam ao variar os controles e relacione com a fórmula.
