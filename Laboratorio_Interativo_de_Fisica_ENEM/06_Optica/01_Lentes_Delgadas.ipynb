## 4. Fórmulas principais
- **1/f = 1/p + 1/p'**
- Use unidades: m.

## 5. Exemplo resolvido
Substitua valores simples na fórmula e confirme o resultado no gráfico.


```python
import numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets

def simular(f,p):
    if p==f:
        print('Objeto no foco: imagem no infinito.'); return
    p_linha=1/(1/f-1/p)
    plt.figure(figsize=(8,3)); plt.axvline(0,color='k'); plt.scatter([-p],[0],label='objeto'); plt.scatter([p_linha],[0],label='imagem')
    plt.xlim(-30,30); plt.yticks([]); plt.grid(); plt.legend(); plt.show()
    print(f'Distância da imagem: {p_linha:.2f} cm')
widgets.interact(simular,f=(5.0,20.0,1.0),p=(6.0,40.0,1.0));
```

## 7. Interpretação física dos resultados
